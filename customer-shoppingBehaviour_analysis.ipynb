```python
#ques- how should we replace the null values?
#-with mean or median, --why mean or why median, --generally choose median over mean because mean can get affected by outliers & that would skew the distribution of our data, -- and we choose median because median is robust to outliers
#we should not do by the method of using overall median of the entire column, because ratings can vary a lot depending on the catergory, so instead we will impute the missing values using the MEDIAN REVIEW RATING WITHIN EACH CATEGORY AND WILL REPLACE THAT WITHIN THE MISSING VALUE(so that the values get replace with it's own median rating rather than using of other column too)
```

```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ', '_')
df=df.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
labels=['Young-adult','Adult','Middle-aged','Senior']
df['age_group']= pd.qcut(df['age'], q=4, labels= labels)
#here qcut splits the ages into four equal-sized groups based on the data distribution and assigns the label we define
```

```python
df[['age','age_group']].head(15)
```

```python
#converting text into number because it makes easy to calculate metrics or compare customers

frequency_mapping= {
    'Fortnightly': 14,
    'Weekly': 7,
    'Monthly': 30,
    'Quarterly': 90,
    'Bi-Weekly': 14,
    'Annually': 365,
    'Every 3 Months': 90
}

df['purchase_frequency_days']= df['frequency_of_purchases'].map(frequency_mapping)
```
